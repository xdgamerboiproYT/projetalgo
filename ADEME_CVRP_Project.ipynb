## 2. Problem Variants: CVRP and CVRPTW

The project proposes two levels of difficulty depending on the constraints considered. Both involve organizing delivery routes for multiple vehicles, but the advanced version introduces additional scheduling requirements.

### 2.1. Level B — Capacitated Vehicle Routing Problem (CVRP)

The CVRP is the standard version of the project. Its objective is to organize delivery routes for several trucks while respecting their maximum loading capacities.

All trucks have the same capacity, and each customer requires a specific quantity of goods.

The main objective is therefore to **minimize the total travel time while ensuring that every customer is served and no vehicle exceeds its capacity**.

### 2.2. Level A — Capacitated Vehicle Routing Problem with Time Windows (CVRPTW)

The CVRPTW is an advanced version of the CVRP. It includes the same vehicle capacity constraints but introduces an additional requirement: **each customer must be delivered within a specific time window**.

This makes the problem more realistic, as some customers may only be available to receive deliveries during certain hours.

As a result, a route that is valid for the CVRP may become impossible for the CVRPTW.

The algorithm must therefore find routes that **minimize total travel time while respecting both vehicle capacities and delivery schedules**.

### 4.2. Mathematical Modeling
Before designing the algorithmic solver, we must establish a rigorous mathematical foundation. We will use graph theory to represent the logistical network, where nodes correspond to the depot and customers, and edges encapsulate the dynamic travel durations between them.

In this section, we will formally define all parameters, decision variables, and operational constraints—such as identical vehicle capacities, time window bounds, and traffic-induced delays. This formulation will culminate in a comprehensive Integer Linear Programming (ILP) model that accurately maps the problem's physical constraints into mathematical equations.

Finally, we will examine the computational complexity of this advanced routing problem (an extension of the classic CVRP known as the Time-Dependent VRPTW). This theoretical study will demonstrate why finding an exact optimal solution becomes exponentially difficult as the network grows, thereby justifying our subsequent use of metaheuristic approaches.


#### 4.2.1. Graph Notation and Sets

We represent the delivery network with a directed graph:

$$
G=(V,E)
$$

The set of vertices contains the depot and the customers:

$$
V=\{v_0,v_1,\ldots,v_n\}
$$

Here, $v_0$ is the depot and $v_i$ is customer $i$ for $i=1,\ldots,n$. The customer set is:

$$
C=V\setminus\{v_0\}
$$

The set of available trucks is represented by:
$$
K=\{1, 2, \ldots, k\}
$$

We use a complete directed graph at the routing level:

$$
E=\{(i,j):i,j\in\{0,\ldots,n\},\ i\neq j\}
$$
And $i$ is the start and $j$ is the end of the path.

### 4.3. Fitness & Objective function

The goal is to minimize the total distance (or total cost) of the journeys across all trucks in the fleet:
$$ \text{Minimize } Z = \sum_{k \in K} \sum_{i \in V} \sum_{j \in V} t_{ij} \times X_{ijk} $$

### 4.4. Constraints modelisation

**Variables and parameters:**<br>
-$T_i$: The decision variable indicating the exact start time of service at customer $i$.
<br>-$X_{ijk}$: The binary decision variable (1 if truck $k$ travels from node $i$ to node $j$, 0 otherwise).
<br>-$[e_i, l_i]$: The time window for customer $i$ ($e_i$ = earliest time, $l_i$ = latest time).
<br>-$t_{ij}$: The travel time between $i$ and $j$.
<br>-$s_i$: The service (unloading) time at customer $i$.
<br>-$q_i$: The demand of customer $i$.
<br>-$Q$: The maximum capacity of a truck.
<br>-$M$: A very large number ("Big M" constant).

**1. Customer Visit Constraint:** 
Each customer must be visited exactly once by exactly one truck:
$$ \sum_{k \in K} \sum_{j \in V} X_{ijk} = 1 \quad \forall i \in C $$

**2. Flow Conservation Constraint:**
If a specific truck $k$ enters customer $h$, that same truck $k$ must leave customer $h$:
$$ \sum_{i \in V} X_{ihk} - \sum_{j \in V} X_{hjk} = 0 \quad \forall h \in C, \forall k \in K $$

**3. Vehicle Capacity Constraint:**
The sum of the demands of all customers visited by a single truck $k$ cannot exceed its maximum capacity $Q$:
$$ \sum_{i \in C} q_i \sum_{j \in V} X_{ijk} \le Q \quad \forall k \in K $$

**4. Time Progression constraint for the route:**
If truck $k$ travels from customer $i$ to customer $j$ (i.e., if $X_{ijk} = 1$), the arrival time at $j$ must be greater than or equal to the start time at $i$, plus the unloading time $s_i$, plus the travel time $t_{ij}$.
$$T_i + s_i + t_{ij} - M(1 - X_{ijk}) \le T_j \quad \forall i, j \in V, \forall k \in K$$

**5. Time window constraint:**
The truck cannot deliver before the opening time $e_i$ nor after the closing time $l_i$. 
$$e_i \le T_i \le l_i \quad \forall i \in V$$
<br><br>

If the truck travels from customer $i$ to customer $j$ (i.e., if $X_{ij} = 1$), the arrival time at $j$ must be greater than or equal to the start time at $i$, plus the unloading time $s_i$, plus the travel time $t_{ij}$.
$$T_i + s_i + t_{ij} - M(1 - X_{ij}) \le T_j \quad \forall i, j \in V$$
<br>
The "Big M" method is an essential mathematical technique used to translate conditional "if-then" logic into a standard linear inequality that Integer Linear Programming solvers can process. By associating a sufficiently large constant M with the binary routing variable, the model actively enforces the time schedule when a truck travels between two specific customers, while mathematically neutralizing the constraint without distorting the overall solution when that route is not taken.


When you ungroup the equations and add the mandatory domain rules with the 3-index formulation (assigning a specific truck $k$), the CVRPTW model breaks down into exactly 14 ILP constraints:

**1. Customer Entry Constraint**
Exactly one truck must enter each customer $j$ from any node $i$ (including the depot).
$$\sum_{k \in K} \sum_{i \in V} X_{ijk} = 1 \quad \forall j \in C$$

**2. Customer Exit Constraint**
Exactly one truck must leave each customer $i$ to any node $j$.
$$\sum_{k \in K} \sum_{j \in V} X_{ijk} = 1 \quad \forall i \in C$$

**3. Depot Departure Limit**
Each individual truck $k$ can leave the depot (Node $0$) at most once. (This naturally ensures no more than $K$ total trucks are used).
$$\sum_{j \in C} X_{0jk} \le 1 \quad \forall k \in K$$

**4. Flow Conservation (Depot and Route Continuity)**
If a specific truck $k$ leaves the depot, it must return to the depot. Furthermore, if truck $k$ enters a customer node $h$, that *same* truck $k$ must leave node $h$.
$$\sum_{i \in V} X_{ihk} - \sum_{j \in V} X_{hjk} = 0 \quad \forall h \in V, \forall k \in K$$

**5. Load Progression (Subtour Elimination)**
If truck $k$ travels from $i$ to $j$, the cumulative load $U_j$ must equal the load at $U_i$ plus the demand $q_j$. (Uses Big M).
$$U_i + q_j - M(1 - \sum_{k \in K} X_{ijk}) \le U_j \quad \forall i, j \in C$$

**6. Vehicle Capacity Bounds**
The cumulative load $U_i$ at any customer must be at least their own demand, but can never exceed the truck's maximum capacity $Q$.
$$q_i \le U_i \le Q \quad \forall i \in C$$
*(Note: Because we now track trucks individually, this can also be expressed directly as: $\sum_{i \in C} q_i \sum_{j \in V} X_{ijk} \le Q \quad \forall k \in K$)*

**7. Time Progression**
If truck $k$ travels from $i$ to $j$, the start time at $j$ must account for the start time at $i$, the service duration $s_i$, and the travel time $t_{ij}$. (Uses Big M).
$$T_i + s_i + t_{ij} - M(1 - X_{ijk}) \le T_j \quad \forall i, j \in V, \forall k \in K$$

**8. Time Window Bounds**
The service start time $T_i$ must fall between the earliest allowed arrival $e_i$ and the latest allowed arrival $l_i$.
$$e_i \le T_i \le l_i \quad \forall i \in V$$

**9. Variable Domains (Integrality & Non-negativity)**
The routing decision variables must be strictly binary, and time/load variables cannot be negative. ILP models are invalid without this declared.
$$X_{ijk} \in \{0, 1\}, \ T_i \ge 0, \ U_i \ge 0 \quad \forall i, j \in V, \forall k \in K$$

**10. Minimum Fleet Size Bound**
Instead of letting the solver guess how many trucks are needed, force it to use a mathematical minimum based on the total demand.
$$ \sum_{k \in K} \sum_{j \in C} X_{0jk} \ge \left\lceil \frac{\sum_{i \in C} q_i}{Q} \right\rceil $$

**11. Depot Operating Hours**
The depot itself usually has an opening time ($e_0$) and a closing time ($l_0$). You must constrain the start time at the depot and ensure all trucks return before the depot closes.
$$ e_0 \le T_0 \le l_0 $$
$$ T_j + s_j + t_{j0} \le l_0 \quad \forall j \in C $$

**12. Self-Loop Elimination**
A mathematically strict model must explicitly state that a truck cannot travel from a customer directly to that exact same customer.
$$ X_{iik} = 0 \quad \forall i \in V, \forall k \in K $$

**13. Incompatible Edges (Time Window Conflicts)**
If it is physically impossible to leave customer $i$ at their earliest possible time, travel to customer $j$, and arrive before customer $j$'s latest possible time, that route should be forbidden for all trucks.
$$ X_{ijk} = 0 \quad \forall k \in K, \text{if } e_i + s_i + t_{ij} > l_j $$

**14. Subtour Elimination for 2 Nodes (2-Cycle Elimination)**
To prevent the solver from testing useless loops where a truck travels from customer $i$ to customer $j$ and immediately back to $i$.
$$ \sum_{k \in K} X_{ijk} + \sum_{k \in K} X_{jik} \le 1 \quad \forall i, j \in C, i \neq j $$

**Sets and Indices**

* **$V$**: The set of all nodes in the network, including the depot and all customers ($V = \{0, 1, 2, ..., N\}$).
* **$C$**: The set of all customer nodes, excluding the depot ($C = V \setminus \{0\}$).
* **$0$**: The index specifically representing the central depot or warehouse.
* **$i, j$**: Indices used to represent specific individual nodes (either the depot or customers) within the network.
* **$N$**: The total number of customers to be served.

**Decision Variables (What the algorithm calculates)**

* **$X_{ij}$**: A binary decision variable. It equals 1 if a truck travels directly from node $i$ to node $j$, and 0 if it does not.
* **$T_i$**: A continuous decision variable representing the exact time a truck begins unloading/service at node $i$.
* **$U_i$**: A continuous decision variable representing the total accumulated weight of goods inside the truck immediately after serving customer $i$.

**Parameters (Constants provided to the algorithm)**

* **$K$**: The maximum number of physical trucks available in your fleet.
* **$Q$**: The maximum weight capacity or volume that a single truck can carry.
* **$M$**: A sufficiently large arbitrary number (Big M) used mathematically to activate or deactivate constraints based on binary routing choices.
* **$q_i$**: The specific demand (weight or quantity of goods) required by customer $i$.
* **$s_i$**: The service time duration required to unload the goods at customer $i$.
* **$t_{ij}$**: The travel time duration required to drive directly from node $i$ to node $j$.
* **$e_i$**: The earliest possible time service can begin at node $i$ (the opening of their time window).
* **$l_i$**: The latest possible time service can begin at node $i$ (the closing of their time window).

Again to prove that our problem is NP, we can show and write a simple checker to verify our proposed solution step by step and see if this checker finishes quickly or in polynomial time without any exponential or factorial calculations.

**the verification steps are as follows:**

**1. verify the customers visited**

the checker will count how many times each customer appears and verifies that every customer is visited once and that every route starts and ends at the depot. checking a list of $N$ customers takes $O(N)$ operations.

**2. verify truck capacity:**

for each trucks route the checker adds the demand $q_{i}$ of every customer on that route. It verifies that this sum is $\leq$ to $Q$ (maximum truck capacity). Summing up all these demands takes $O(N)$ operations.

**3. verify time windows and travel logic**

the checker will see the timeline of each route. for a truck traveling from $i$ to customer $j$ it will calculate if the arrival time at $j$ is possible (arrival at $i$ + service time + travel time) it will then check if the arrival time is within the time window $[e_{j},l_{j}]$. doing basic additions and bound checking for $N$ stops takes $O(N)$ operations.

**4. verify the budget**

Finally, the checker will add all the travel times of every single road used and then see if this total is $\leq$ to our budget B. Summing the travel time takes $O(N)$ operations.


**Conclusion**

Since every single step requires simple arithmetic( additions, comparaisons) and loops through N customers, Therefore the whole verification process runs in polynomial time.

Because any solution to the CVRPTW decision problem can be mathematically verified in polynomial time we can conclude that the problem belongs to the NP class.

### 5.2. Hamiltonian Cycle to TSP

Given $G = (V, E)$, we construct a complete graph $G' = (V, E')$ with the same set of vertices $V$.

Since $G'$ is complete, there is an edge between every pair of vertices $i$ and $j$ ($E' = \{(i,j) \mid i, j \in V, i \neq j\}$).

We define the cost function $c: E' \rightarrow \mathbb{N}$ for each edge $(i,j)$ of $G'$ as follows:<br><br>

$$c_{ij} = \begin{cases} 1 & \text{if } (i,j) \in E \text{ (The edge exists in the original graph } G) \\ 2 & \text{if } (i,j) \notin E \text{ (The edge does not exist in the original graph. } G) \end{cases}$$
<br><br>
Finally, we set the total cost bound $B = \vert{}V\vert{}$ (where $\vert{}V\vert{}$ is the number of vertices, that is, the number of stops on the tour).

This transformation involves checking whether the original edges exist in order to allocate costs.
<br>
It runs in $O(\vert{}V\vert{}^2)$ time, which is indeed a polynomial time.
<br><br>
Step 2 : Proof of Equivalence (Necessary and Sufficient Condition)
We must now prove that the HC instance is true if and only if the constructed TSP instance is true.
We must now prove that the HC instance is true if and only if the constructed instance of TSP is true.
<br><br>
(Direct direction: Hamiltonian Cycle $\implies$ TSP)
<br>
Suppose there exists a Hamiltonian cycle in $G$.
<br>
This cycle is a sequence of $\vert{}V\vert{}$ vertices connected by $\vert{}V\vert{}$ edges belonging to $E$.
<br>
In the complete graph $G'$, if the traveling salesman follows exactly this same cycle, he will travel only along edges with a cost of $c_{ij} = 1$.
The total cost of this tour will therefore be $\sum c_{ij} = \vert{}V\vert{} \times 1 = \vert{}V\vert{}$.
<br>
Since the cost is equal to $B$ ($\vert{}V\vert{} \le B$), the TSP decision instance is verified.
<br><br>
(Indirect direction: TSP $\implies$ HC) 
<br>
Suppose there exists in $G'$ a traveling salesman tour whose total cost is less than or equal to $B = \vert{}V\vert{}$.
A valid tour visits each vertex exactly once and returns to its starting point; therefore, it consists of exactly $\vert{}V\vert{}$ edges.
Since the minimum cost of an edge in $G'$ is $1$, the minimum possible cost for $\vert{}V\vert{}$ edges is exactly $\vert{}V\vert{}$.

For the total cost not to exceed the bound $B = \vert{}V\vert{}$, it is essential that the traveler has traveled only along edges with cost $c_{ij} = 1$.
<br><br>
By the definition of our cost function, an edge has a cost of $1$ if and only if it belongs to $E$. The tour therefore uses only edges that exist in the original graph $G$, thus forming a Hamiltonian cycle in $G$.

We will prove that the CVRPTW is NP-Hard by constructing a polynomial reduction from the Traveling Salesman Problem (TSP), whose decision version is NP-Complete. 
<br><br>
The decision version of the TSP is as follows: 
<br><br>
Given a weighted complete graph $G = (V, E)$, an edge cost function $c : E \rightarrow \mathbb{R}^+$, and a bound $B$, does there exist a Hamiltonian cycle (a tour visiting each vertex exactly once) whose total cost is less than or equal to $B$? The decision version of the CVRPTW is defined as follows: Given a weighted complete graph $G' = (V', E')$, a depot $0$, customers $C$, a demand $q_i$, a time window $[e_i, l_i]$, and a service time $s_i$ for each customer $i$, travel times $t_{ij}$ between nodes, a fleet of $k$ vehicles with capacity $Q$, and a cost bound $B'$, does there exist a set of valid routes that respect the capacities and time windows, visit each customer once, and have a total cost of $\le B'$?

**Step 1: Constructing the CVRPTW Instance (Transformation)**
Given any TSP instance defined by $(G = (V, E), c, B)$ with $n = \vert{}V\vert{}$ vertices, we construct a CVRPTW instance as follows:
<br><br>
Graph and Depot: The graph remains the same, $G' = G$, and the cost matrix remains $c' = c$. We choose an arbitrary vertex $v_0 \in V$ to be the depot. The other vertices form the set of customers $C$. 
<br><br>
Vehicle Fleet: We limit the fleet to a single vehicle, $k = 1$. 
<br><br>
Capacity and Demands (CVRP Relaxation): We negate the effect of capacity by setting each customer’s demand to $q_i = 1$ and the vehicle’s capacity to $Q = n$. The vehicle can thus transport all demands in a single trip. 
<br><br>
Time and Windows (VRPTW Relaxation): We eliminate the effect of time constraints by setting: <br>
Window opening times to the initial time: $e_i = 0, \forall i \in V$. <br>
Window closing times to infinity: $l_i = +\infty, \forall i \in V$. <br>
Service times to zero: $s_i = 0, \forall i \in V$. <br>
Travel times to zero: $t_{ij} = 0, \forall (i,j) \in E$. (Or optionally $t_{ij} = c_{ij}$; this has no effect since the windows close at $+\infty$).
<br><br>
Cost bound: We set $B' = B$. 
<br><br>
This transformation consists of initializing the CVRPTW parameters based on the TSP instance. It requires a number of operations proportional to the number of vertices and edges, so it is performed in polynomial time.


## 6. Problem Perimeter and Constraints

### 6.1. What the Model Includes

We consider one depot, at most $k$ identical trucks, known positive customer demands, a single delivery per customer, customer time windows, known service durations, and travel times that depend on departure time. Every active truck performs **one route**, starting and ending at the depot. Trucks may wait when they arrive early, and some available trucks may remain unused.

The objective is to **minimize the sum of driving durations across all trucks**. Waiting and service time influence feasibility, but they are not added to this particular objective. This distinction matters: minimizing driving time is not the same as minimizing the final completion time of all routes.

### 6.2. What the Model Does Not Include

We do not model split deliveries, multiple depots, different truck capacities, vehicle breakdowns, stochastic traffic, driver rest rules, loading times at the depot, or fuel/emission coefficients. These could be studied in future extensions.

We assume all time windows and travel-time forecasts are known before planning begins. The planning horizon is finite and every selected trip must finish within it.

### 6.3. Feasible Solution

A delivery plan is feasible only if every customer is served by exactly one truck, the load of each truck is at most $Q$, each active truck departs from and returns to the depot, and every service starts within the customer's time window. Consecutive visits must also respect travel and service durations.

### 7.1. Possible Solution Approaches

As explained earlier, the CVRPTW is an NP-hard problem, so finding the optimal solution can become difficult when the number of customers increases.

We plan to use **Hill Climbing** as our main heuristic. This method starts with an initial solution and gradually improves it by making small changes to the delivery routes.

We also plan to use **Multi-Start**, which means running Hill Climbing several times with different initial solutions. This could help us find better routes and avoid depending on a single starting point.

For the metaheuristic, we are considering either **Tabu Search** or **Simulated Annealing**. Both methods could help us escape local optima and explore more possible solutions.

Our goal is to find efficient delivery routes while respecting vehicle capacities and time windows.

### 7.2. Working Hypotheses

We have identified several hypotheses to test during the project:

- **H1 — Hill Climbing:** We expect Hill Climbing to improve an initial solution quickly, although it may get stuck in a local optimum.
- **H2 — Multi-Start:** Running Hill Climbing from different starting solutions could help us find better routes.
- **H3 — Metaheuristics:** We believe that Tabu Search or Simulated Annealing could find better solutions than basic Hill Climbing by exploring more possibilities.
- **H4 — Problem size:** We expect the problem to become harder as the number of customers increases and delivery time windows become more restrictive.
- **H5 — Traffic conditions:** We believe that traffic variations may change the most efficient routes depending on departure times.
- H6 — Multi-Start with Metaheuristics: We believe that applying Multi-Start to Tabu Search or Simulated Annealing could improve solution quality by exploring different starting points. However, we expect the improvement to be smaller than with Hill Climbing, since these metaheuristics already have ways to escape local optima.
